```python
# Instale as bibliotecas necessárias
!pip install requests
!pip install beautifulsoup4

import requests
from bs4 import BeautifulSoup

```

```python
# URL do site de exemplo hospedado no GitHub Pages
url = 'https://books.toscrape.com/'

# Fazendo a requisição HTTP para obter o conteúdo da página
response = requests.get(url)
```

```python
response
```

```python
if response.status_code == 200:
    print("Conexão bem-sucedida!")

    # Criação do objeto BeautifulSoup para parsing do HTML
    soup = BeautifulSoup(response.text, 'html.parser')

    # Encontrando todos os produtos na página
    livros = soup.find_all('article', class_='product_pod')

    lista_livros = []
    total_preco = 0
    contagem_livros = 0

    for produto in livros:
        nome = produto.h3.a['title']
        preco = produto.find('p', class_='price_color').text
        valorajustado = float(preco.replace('Â£', '').replace('£', ''))
        descricao = produto.find('p', class_='instock availability').text.strip()

        # Lista
        lista_livros.append({'nome': nome, 'preco': valorajustado, 'descricao': descricao})

        # Contagem
        total_preco += valorajustado
        contagem_livros += 1

        print("Nome:", nome)
        print("Preço:", valorajustado)
        print("Descrição:", descricao)
        print("-" * 20)

else:
    print("Erro ao acessar o site.")
```

```python

    # 1. Preço médio dos livros: Mostra o preço médio de todos os livros listados.
preco_medio = total_preco / contagem_livros
print("\nPreço médio dos livros:", round(preco_medio, 2))

    # 2. Contagem de livros disponíveis: Informa quantos livros estão disponíveis no estoque.
print("Contagem de livros disponíveis:", contagem_livros)

    # 3. Livros mais caros: Exibe os 3 livros com os preços mais altos na lista.
livros_ordenados = sorted(lista_livros, key=lambda x: x['preco'], reverse=True)
print("\nLivros mais caros:")
for livro in livros_ordenados[:3]:
  print("Nome:", livro['nome'], "| Preço:", livro['preco'])
```
