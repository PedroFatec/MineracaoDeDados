Importação de Bibliotecas

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

Examinação de Dataset

```python
df = pd.read_csv("/content/car_price_prediction_.csv")
df.head()
```

```python
# Verificar a presença de valores nulos
df.isnull().sum()
```

```python
# Substituir valores nulos nas colunas numéricas pela média
df['Price'].fillna(df['Price'].mean(), inplace=True)

# Substituir valores nulos nas colunas categóricas por um valor padrão, como 'Desconhecido'
df['Brand'].fillna('Desconhecido', inplace=True)
```

```python
# Converter coluna categórica para tipo 'category'
df['Brand'] = df['Brand'].astype('category')

# Converter coluna numérica para tipo numérico
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
```

```python
# Estatísticas descritivas para variáveis numéricas
df.describe()


```

```python
# Frequência de valores para variáveis categóricas
df['Brand'].value_counts()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Histograma para a variável numérica
sns.histplot(df['Price'], bins=10, kde=True)
plt.title('Distribuição da Variável Numérica')
plt.show()
```

```python
# Boxplot para a variável numérica
sns.boxplot(x=df['Price'])
plt.title('Boxplot da Variável Numérica')
plt.show()
```

```python
# Gráfico de barras para variável categórica
sns.countplot(x='Brand', data=df)
plt.title('Distribuição de Categorias')
plt.show()
```

```python
# Gráfico de dispersão entre duas variáveis numéricas
sns.scatterplot(x='Price', y='Engine Size', data=df)
plt.title('Correlação entre Variáveis Numéricas')
plt.show()
```
